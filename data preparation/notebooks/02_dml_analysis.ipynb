# Aniket — Week 3 DML Analysis

This notebook runs the reusable pipeline. Input data are simulated; estimates are exploratory and not validated causal findings.

```python
from pathlib import Path
import sys
import pandas as pd

PROJECT = Path('..').resolve()  # notebook lives in notebooks/
sys.path.insert(0, str(PROJECT / 'src'))
from dml_pipeline import load_panel, estimate_dml

DATA_PATH = PROJECT / 'data' / 'processed' / 'cleaned regional market transactions.csv'
df = load_panel(DATA_PATH)
print('Shape:', df.shape)
print('Markets:', df['market_id'].nunique())
print('Products:', df['product_id'].nunique())
print(df.groupby('treatment')[['gross_profit', 'revenue', 'units_sold']].agg(['count', 'mean']))
```

## Run the three exploratory outcomes

The standard error output is IID-only and must not be used as final panel inference. Review overlap, treatment assignment, and causal assumptions before interpreting estimates.

```python
results = []
for outcome in ['gross_profit', 'revenue', 'units_sold']:
    result, oof = estimate_dml(df, outcome, n_splits=5)
    results.append(result)
    print(outcome, result['dml_point_estimate_exploratory'], result['overlap_flag'], result['causal_status'])

results_df = pd.DataFrame(results)
display(results_df[['outcome', 'n_rows_used', 'dml_point_estimate_exploratory', 'overlap_flag', 'causal_status']])
```
